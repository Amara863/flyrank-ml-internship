## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Directly load from GitHub repo
url = "https://raw.githubusercontent.com/Amara863/flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = len(df)
declining_count = df["trend_direction"].str.lower().eq("down").sum()
declining_pct = (declining_count / total_pages) * 100

stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]

print(f"1. Total Pages in Dataset: {total_pages}")
print(f"2. Declining Pages Rate: {declining_pct:.1f}% ({declining_count} pages)")
print(f"3. High-Value Stale Pages (>=180 days old, >=500 imp): {len(stale_visible)} pages")
```
